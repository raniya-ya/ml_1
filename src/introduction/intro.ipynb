**4. Think about what the difference is between multiclass and multilabel.**

Мультиклассовая классификация - в этой задаче обьект относится только к одному классу из нескольких. Например, определение животного: либо собака, либо кошка и тд.

Мультилейбл - обьект может принадлежать нескольким классам одновременно. Пример: фильм может иметь несколько жанров (комедия и боевик).

**5. Is an example case with house prices from the theory a classification of a regression problem? Is it possible to reduce the regression problem to classification?**


Пример с ценами на дома является задачей регрессии, потому что мы предсказываем точное число. 
Регрессию можно свести к задаче классификации путем дискретизации. То есть можно, например, разбить цены на категории (дешево, дорого). Но таким подходом теряется точность, получаем лишь диапазон, в которую попадает цена дома.
