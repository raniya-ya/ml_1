```python
df_train = pd.read_csv("../data/df_no_outliers.csv")
df_test = pd.read_csv("../data/df_test_no_outliers.csv")
```

**Results table**


Create two empty Pandas DataFrames with columns 'model', 'train', 'test'. Let's call the first one result_MAE and the second one result_RMSE. We will fill these tables with the results of the models.

```python
result_MAE = pd.DataFrame(columns = ['model','train','test'])
result_RMSE = pd.DataFrame(columns = ['model','train','test'])
```


**Linear Regression**

Initialize linear regression from sklearn with no parameters.

Fit your model and make predictions on training and test features. Save it as new columns in data.

Compute MAE (Mean Absolute Error) on training and test targets.

Calculate RMSE (Root Mean Square Error) on training and test objectives.

Insert your metrics into tables result_MAE and result_RMSE with model name 'linear_regression'.


```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, root_mean_squared_error
```

```python
x_train = df_train[["bathrooms", "bedrooms"]]
y_train = df_train["price"]

x_test = df_test[['bathrooms','bedrooms']]
y_test = df_test['price']
```

```python
linear_regression = LinearRegression()
linear_regression.fit(x_train, y_train)
```

```python
y_pred_train_lr = linear_regression.predict(x_train)
y_pred_test_lr = linear_regression.predict(x_test)
```

```python
mae_train_lr = mean_absolute_error(y_train, y_pred_train)
mae_test_lr = mean_absolute_error(y_test, y_pred_test)
```

```python
mae_train_lr
```

```python
mae_test_lr
```

```python
rmse_train_lr = root_mean_squared_error(y_train, y_pred_train)
rmse_test_lr = root_mean_squared_error(y_test, y_pred_test)
```

```python
rmse_train_lr
```

```python
rmse_test_lr
```

```python
result_MAE.loc[0] = ['linear_regression', mae_train_lr, mae_test_lr]
result_RMSE.loc[0] = ['linear_regression', rmse_train_lr, rmse_test_lr]
```

**Decision Tree**

Initialize decision tree regressor from sklearn with fixed random_state=21.
    
Fit it to train features and train target and make prediction on train and test features. Save it as a new column in data.
    
Compute MAE (Mean Absolute Error) on train and test targets.
    
Compute RMSE (Root Mean Square Error) on train and test targets.
    
Insert your metrics into tables result_MAE and result_RMSE with model name 'decision_tree'.


```python
from sklearn.tree import DecisionTreeRegressor
```

```python
tree = DecisionTreeRegressor(random_state=21)
```

```python
tree.fit(x_train, y_train)
```

```python
y_pred_train_dt = tree.predict(x_train)
y_pred_test_dt = tree.predict(x_test)
```

```python
mae_train_dt = mean_absolute_error(y_train, y_pred_train_dt)
mae_test_dt = mean_absolute_error(y_test, y_pred_test_dt)
```

```python
rmse_train_dt = root_mean_squared_error(y_train, y_pred_train_dt)
rmse_test_dt = root_mean_squared_error(y_test, y_pred_test_dt)
```

```python
result_MAE.loc[1] = ['decision_tree', mae_train_dt, mae_test_dt]
result_RMSE.loc[1] = ['decision_tree', rmse_train_dt, rmse_test_dt]
```

**Naive Models**

Calculate the mean and median of 'price' on the training and test data and create a column with these values.

Calculate the MAE on the training and test targets between your target and the calculated mean and median.

Calculate the RMSE on the training and test targets between your target and the calculated mean and median.

Insert your metrics into tables result_MAE and result_RMSE with model names 'naive_mean' and 'naive_median'.

```python
mean_price = y_train.mean()
median_price = y_train.median()
```

```python
y_pred_train_mean = [mean_price] * len(y_train)
y_pred_test_mean = [mean_price] * len(y_test)
```

```python
y_pred_train_median = [median_price] * len(y_train)
y_pred_test_median = [median_price] * len(y_test)
```

```python
mae_train_mean = mean_absolute_error(y_train, y_pred_train_mean)
mae_test_mean = mean_absolute_error(y_test, y_pred_test_mean)

mae_train_median = mean_absolute_error(y_train, y_pred_train_median)
mae_test_median = mean_absolute_error(y_test, y_pred_test_median)
```

```python
rmse_train_mean = root_mean_squared_error(y_train, y_pred_train_mean)
rmse_test_mean = root_mean_squared_error(y_test, y_pred_test_mean)

rmse_train_median = root_mean_squared_error(y_train, y_pred_train_median)
rmse_test_median = root_mean_squared_error(y_test, y_pred_test_median)
```

```python
result_MAE.loc[2] = ['naive_mean', mae_train_mean, mae_test_mean]
result_RMSE.loc[2] = ['naive_mean', rmse_train_mean, rmse_test_mean]
```

```python
result_MAE.loc[3] = ['naive_median', mae_train_median, mae_test_median]
result_RMSE.loc[3] = ['naive_median', rmse_train_median, rmse_test_median]
```
